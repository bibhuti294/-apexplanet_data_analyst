DATA CLEANING AND PREPROCESSING

```python
df.isnull().sum()
```

```python
(df['TotalCharges'] == ' ').sum()
```

```python
df['TotalCharges'] = pd.to_numeric(
    df['TotalCharges'],
    errors='coerce'
)
```

```python
df.isnull().sum()
```

```python
df['TotalCharges'].fillna(
    df['TotalCharges'].median(),
    inplace=True
)
```

```python
df.duplicated().sum()
```

```python
df.columns = df.columns.str.lower()
```

Converting to Categorical Columns

```python
cat_cols = df.select_dtypes(
    include='object'
).columns

for col in cat_cols:
    df[col] = df[col].astype('category')
```

```python
numeric_cols = ['tenure', 'monthlycharges', 'totalcharges']

for col in numeric_cols:
    sns.boxplot(x=df[col])
    plt.title(col)
    plt.show()
```

Outlier Analysis:
Boxplots and IQR analysis were performed on tenure, MonthlyCharges, and TotalCharges.
No significant erroneous outliers were identified. Extreme values correspond to genuine long-term and premium customers. Therefore, no outlier treatment was applied.

## Data Cleaning Log

1. Loaded the IBM Telco Customer Churn dataset.
2. Examined the dataset dimensions, columns, and data types.
3. Identified hidden missing values (blank spaces) in the `TotalCharges` column.
4. Converted `TotalCharges` from object to numeric data type.
5. Imputed 11 missing values in `TotalCharges` using the median.
6. Verified that the dataset contains no duplicate records.
7. Standardized all column names to lowercase for consistency.
8. Converted categorical columns to the `category` data type where appropriate.
9. Performed outlier analysis using boxplots and IQR inspection.
10. Retained extreme values as they represent valid business cases (e.g., long-term and high-value customers) rather than data errors.
